```python

print("## ワイド分散ベーススコア")
print("wide_score_adj = EV / (E × sqrt(P))")
print("")
print("旧スコア: EV / P (E を無視 → 分散を考慮しない)")
print("新スコア: EV / (E × sqrt(P)) (分散リスクを調整)")
print("")
print("Var_proxy = E × sqrt(P) は払戻の分散を近似")
print("  → 高オーズ・低確率のペアほどスコアが下がる")
print("  → より安定したベット選択が可能")
```

```python
print("""
## スコア比較: 旧スコア vs 新スコア

例: EV = 2.0, P = 0.1, E = 20.0
  旧スコア = 2.0 / 0.1 = 20.0
  新スコア = 2.0 / (20.0 × sqrt(0.1)) = 2.0 / 6.32 = 0.316

例: EV = 1.5, P = 0.3, E = 5.0
  旧スコア = 1.5 / 0.3 = 5.0
  新スコア = 1.5 / (5.0 × sqrt(0.3)) = 1.5 / 2.74 = 0.548

新スコアでは低リスクのベットが優先される。
""")
```

```python
print("""
## DD (ドローダウン) 改善効果

新スコアの主要効果:
1. 最大DDの抑制
   - 高オーズ馬のスコアが下がる
   - 不的中時の連続損失を軽減

2. シャープレシオの改善
   - 分散が小さいベットを優先
   - 投資額の変動が小さくなる

3. 長期ROI の安定化
   - 分散調整で過剤ベットを回避
   - 結果的に ROI の標準偏差が下がる
""")
```
