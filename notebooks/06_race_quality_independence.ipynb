```python
print("""
## 投資率 vs ROI のトレードオフ

Screener の収容度を調整すると、投資率とROIの関係が変わる。

収容度を厳しく (投資率↓):
  - メリっとある高品質レースだけに投資
  - ROI は向上するが、投資回数が減る
  - 統計的な有意性が下がる可能性

収容度を緩める (投資率↑):
  - より多くのレースに投資
  - ROI は下がるが、投資回数が増える
  - 統計的な有意性は向上

最適な収容度は、ROI×投資率 (expected profit) を最大化する点。
""")
```
